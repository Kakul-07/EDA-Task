### Numerical Columns

```python
numerical_columns = df.select_dtypes(include=np.number).columns
print("Numerical Columns:",numerical_columns)
```

### Categorical Columns

```python
categorical_columns = df.select_dtypes(include=["object","str"]).columns
print("\nCategorical Columns:",categorical_columns)
```

### Distribution of numerical features

```python
df[numerical_columns].hist(figsize=(12, 10), bins=20)
plt.tight_layout()
plt.show()
```

### Boxplot of numerical columns

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[numerical_columns]) 
plt.title("Boxplots of Numerical Features")
plt.show()
```

### Age Distribution

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Age", bins=20, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

### Heart disease distribution

```python
plt.figure(figsize=(6, 6))
df["HeartDisease"].value_counts().plot( kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Heart Disease Distribution")
plt.show()
```

### Heart disease percentage

```python
heart_disease_percentage = df["HeartDisease"].value_counts(normalize=True) * 100
print(heart_disease_percentage)
```

### Gender distribution

```python
plt.figure(figsize=(6, 6))
df["Sex"].value_counts().plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Gender Distribution")
plt.ylabel("")
plt.show()
```

### Heart disease by sex

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Sex", hue="HeartDisease")
plt.title("Heart Disease by Sex")
plt.xlabel("Sex")
plt.ylabel("Number of Patients")
plt.show()
```

### Chest pain distribution

```python
plt.figure(figsize=(6, 6))
df["ChestPainType"].value_counts().plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Chest Pain Type Distribution")
plt.ylabel("")
plt.show()
```

### Chest pain vs heart disease

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="ChestPainType", hue="HeartDisease")
plt.title("Chest Pain Type vs Heart Disease")
plt.xlabel("Chest Pain Type")
plt.ylabel("Number of Patients")
plt.show()
```

### Exercise induced angina vs heart disease

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="ExerciseAngina", hue="HeartDisease")
plt.title("Exercise Angina vs Heart Disease")
plt.xlabel("Exercise Angina")
plt.ylabel("Number of Patients")
plt.show()
```

### ST slope vs heart disease

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="ST_Slope", hue="HeartDisease")
plt.title("ST Slope vs Heart Disease")
plt.xlabel("ST Slope")
plt.ylabel("Number of Patients")
plt.show()
```

###  Resting ECG vs heart disease

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="RestingECG", hue="HeartDisease")
plt.title("Resting ECG vs Heart Disease")
plt.xlabel("Resting ECG")
plt.ylabel("Number of Patients")
plt.show()
```

### Age vs Heart Disease

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="HeartDisease", y="Age")
plt.title("Age vs Heart Disease")
plt.xlabel("Heart Disease")
plt.ylabel("Age")
plt.show()
```
